# Part A: Why Pandas?

NumPy is great for numbers, but real-world data usually looks like a **table** — rows and columns, with column names, and often a mix of text and numbers together (like a student's Name **and** their Score).

**Pandas** is a Python library built for exactly this. It works like a spreadsheet or a small database, inside Python.

Its main object is called a **DataFrame**.

```python
import pandas as pd

print("Pandas is ready! Version:", pd.__version__)
```

- `.shape` → shows `(number of rows, number of columns)`
- `.columns` → shows the column names
- `.describe()` → shows summary statistics (count, mean, min, max, etc.) for numeric columns

```python
import pandas as pd

df = pd.read_csv("students.csv")

print("First 5 rows:")
print(df.head())
```

# Part D: Handling Missing Data

Real datasets often have missing values (empty cells). Pandas shows these as `NaN` (Not a Number).

`df.fillna(value)` replaces every missing value with the value you give it.

```python
import pandas as pd
import numpy as np

# Let's make a small copy with a missing value, just to see how this works
df_demo = pd.read_csv("students.csv").copy()
df_demo.loc[5, "Department"] = np.nan   # pretend Sana's score wasn't recorded

print("Before fillna (notice the NaN):")
print(df_demo)

df_demo_filled = df_demo.fillna("Yet Not Decided")
print("\nAfter fillna(0):")
print(df_demo_filled)
```

# Part E: Selecting and Filtering

### Selecting specific columns
Use double square brackets with the column names you want: `df[["Col1", "Col2"]]`

### Filtering rows by a condition
Write a condition inside square brackets, e.g. `df[df["Score"] > 80]`. Pandas keeps only the rows where the condition is `True`.

```python
import pandas as pd

df = pd.read_csv("students.csv")

# Selecting only the Name and Score columns
df[["Name", "Score"]]
```

```python
import pandas as pd

df = pd.read_csv("students.csv")

# Filtering: keep only students with Score > 80
df[df["Age"] > 23]
```

```python
import pandas as pd

df = pd.read_csv("students.csv")

# Filtering: keep only students in the Computer Science department
df[df["Department"] == "Data Science"]
```

```python
# Filtering: Based on Names
df[df["Name"] == "Zain"]
```

# Part F: Adding and Removing Columns

Adding a new column is as simple as assigning it a value, often computed from other columns.

Removing a column uses `.drop(columns=["ColumnName"])`.

```python
import pandas as pd

df = pd.read_csv("students.csv")

# Adding a new column: Pass/Fail based on Score
df["Result"] = df["Score"].apply(lambda s: "Pass" if s >= 60 else "Fail")
df
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df["Result"] = df["Score"].apply(lambda s: "Pass" if s >= 60 else "Fail")

# Removing the Result column again
df_dropped = df.drop(columns=["Result"])
df_dropped
```
